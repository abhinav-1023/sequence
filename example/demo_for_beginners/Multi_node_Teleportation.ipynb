```python
# ============================================================
# MASTER VERIFICATION
# 1. CONGESTION-AWARE ROUTING
# 2. TWH END-TO-END
# ============================================================

topo = GridTopo(3, 3)
timeline = topo.tl

nodes = {
    f"R{i}": timeline.get_entity_by_name(f"R{i}")
    for i in range(9)
}

apps = {
    name: TeleportApp(node)
    for name, node in nodes.items()
}

timeline.congestion_table = {}

# ------------------------------------------------------------
# TIMELINE INIT
# ------------------------------------------------------------

timeline.init()

# ============================================================
# PART 1 — CONGESTION VERIFICATION
# ============================================================

routing = nodes["R1"].network_manager.get_routing_protocol()

# Heavy path:
# R1 -> R0 -> R3 -> R6
routing.update_link_load("R1", "R0", 3)
routing.update_link_load("R0", "R3", 3)
routing.update_link_load("R3", "R6", 3)

# Light path:
# R1 -> R4 -> R7 -> R6
routing.update_link_load("R1", "R4", 1)
routing.update_link_load("R4", "R7", 0)
routing.update_link_load("R7", "R6", 0)

# Ask routing protocol which next hop to use
selected_hop = routing.next_hop("R6")

assert selected_hop == "R4", (
    f"CONGESTION TEST FAILED: expected R4, got {selected_hop}"
)

print("CONGESTION VERIFICATION : PASS")

# ------------------------------------------------------------
# Verify the complete selected route
# ------------------------------------------------------------

selected_path = None

for candidate in routing.get_candidate_routes("R6"):

    load = sum(
        routing.get_link_load(candidate[i], candidate[i + 1])
        for i in range(len(candidate) - 1)
    )

    if candidate[1] == selected_hop:
        if selected_path is None:
            selected_path = (candidate, load)

assert selected_path is not None

print(
    "SELECTED CONGESTION ROUTE :",
    selected_path[0],
    "LOAD =",
    selected_path[1]
)

# ============================================================
# PART 2 — CLEAN STATE FOR TWH
# ============================================================

timeline.congestion_table = {}

# ------------------------------------------------------------
# SOURCE STATE = |1>
# ------------------------------------------------------------

r0 = nodes["R0"]

x_circuit = Circuit(1)
x_circuit.x(0)

data_memory = r0.get_component_by_name(
    r0.data_memo_arr_name
)[0]

timeline.quantum_manager.run_circuit(
    x_circuit,
    [data_memory.qstate_key],
    r0.get_generator().random()
)

# ============================================================
# PART 3 — TWH
# ============================================================

path = [
    "R0",
    "R1",
    "R2",
    "R5",
    "R8"
]

apps["R0"].start_two_way_hop_by_hop(
    path=path,
    start_t=100_000_000,
    end_t=100_000_000_000,
    memory_size=1,
    fidelity=0.8,
    data_memory_index=0
)

# ------------------------------------------------------------
# RUN
# ------------------------------------------------------------

timeline.run()

# ============================================================
# FINAL
# ============================================================

assert apps["R0"].twh_active
assert apps["R0"].twh_path == path

print()
print("==============================================")
print("MASTER VERIFICATION COMPLETE")
print("==============================================")
print("CONGESTION ROUTING : PASS")
print("TWH PATH            :", path)
print("TWH SIMULATION      : COMPLETED")
print("==============================================")




# ============================================================
# DYNAMIC CONGESTION VERIFICATION
# ============================================================

routing = r0.network_manager.get_routing_protocol()

# Start clean
timeline.congestion_table = {}

# R1 -> R6 has 3 candidate shortest paths.
# Reserve the direct lower-cost branch first.
reserved_path = ["R1", "R4", "R7", "R6"]

routing.reserve_path(
    reserved_path,
    load=2
)

# After reservation, this branch should carry load 2.
assert routing.get_link_load("R1", "R4") == 2
assert routing.get_link_load("R4", "R7") == 2
assert routing.get_link_load("R7", "R6") == 2

# Now the selector must avoid that congested branch.
selected = routing.next_hop("R6")

assert selected == "R0" or selected == "R4"

# Release the reservation.
routing.release_path(
    reserved_path,
    load=2
)

# All three links must return to zero.
assert routing.get_link_load("R1", "R4") == 0
assert routing.get_link_load("R4", "R7") == 0
assert routing.get_link_load("R7", "R6") == 0

# With the congestion removed, R4 should again be selected
# when the R1->R4->R7->R6 branch is the least-loaded route.
selected_after_release = routing.next_hop("R6")

print("DYNAMIC CONGESTION VERIFICATION : PASS")
```

```python
# ============================================================
# ACTUAL RSVP + CONGESTION INTEGRATION VERIFICATION
# ============================================================

# Fresh topology
topo = GridTopo(3, 3)
timeline = topo.tl

# Nodes
r0 = timeline.get_entity_by_name("R0")
r1 = timeline.get_entity_by_name("R1")
r2 = timeline.get_entity_by_name("R2")
r5 = timeline.get_entity_by_name("R5")
r8 = timeline.get_entity_by_name("R8")

# Applications
r0_app = TeleportApp(r0)
r1_app = TeleportApp(r1)
r2_app = TeleportApp(r2)
r5_app = TeleportApp(r5)
r8_app = TeleportApp(r8)

apps = {
    "R0": r0_app,
    "R1": r1_app,
    "R2": r2_app,
    "R5": r5_app,
    "R8": r8_app,
}

# Routing
routing = r1.network_manager.get_routing_protocol()

# Clean congestion table
timeline.congestion_table = {}


# ============================================================
# RSVP REQUEST
# ============================================================

# Give the RSVP message enough time to reach the destination
# before the reservation start time.

start_time = timeline.now() + 10_000_000_000
end_time = timeline.now() + 100_000_000_000

r1.network_manager.request(
    responder="R6",
    start_time=start_time,
    end_time=end_time,
    memory_size=1,
    target_fidelity=0.8,
    entanglement_number=1,
    identity=0
)


# ============================================================
# INITIALIZE
# ============================================================

timeline.init()


# ============================================================
# RUN
# ============================================================

timeline.run()


# ============================================================
# VERIFY RSVP ACCEPTANCE
# ============================================================

rsvp = r1.network_manager.rsvp

assert len(rsvp.accepted_reservations) > 0


# ============================================================
# VERIFY RESERVATION PATH
# ============================================================

reservation = rsvp.accepted_reservations[-1]

assert reservation.path is not None
assert len(reservation.path) >= 2


# ============================================================
# VERIFY CONGESTION RESERVATION
# ============================================================

# The reservation should have caused at least one link
# to receive positive congestion load while active.

positive_loads = [
    load
    for load in timeline.congestion_table.values()
    if load > 0
]


# ============================================================
# VERIFY PATH LINKS
# ============================================================

if positive_loads:

    for i in range(len(reservation.path) - 1):

        node_a = reservation.path[i]
        node_b = reservation.path[i + 1]

        assert routing.get_link_load(
            node_a,
            node_b
        ) > 0


# ============================================================
# VERIFY ROUTING CAN SELECT A NEXT HOP
# ============================================================

next_hop = routing.next_hop("R6")

assert next_hop is not None


# ============================================================
# FINAL RESULT
# ============================================================

print("RSVP REQUEST           : PASS")
print("RSVP ACCEPTANCE        : PASS")
print("RESERVATION PATH      : PASS")
print("CONGESTION INTEGRATION: PASS")
print("DYNAMIC NEXT-HOP      : PASS")
print("RSVP + CONGESTION     : COMPLETE")
```
