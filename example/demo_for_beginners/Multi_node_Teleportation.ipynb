```python

topo = GridTopo(3,3)

print("Total routers:", len(topo.nodes["QuantumRouter"]))

print("\nRouters:")
for node in topo.nodes["QuantumRouter"]:
    print(
        node.name,
        "type =", type(node).__name__
    )
    
print("\n===== QUANTUM CHANNELS =====")

for qc in topo.qchannels:
    print(qc.name, ":", qc.sender.name, "->", qc.receiver)
print("\n===== CLASSICAL CHANNELS =====")

for cc in topo.cchannels:
    print(cc.name, ":", cc.sender.name, "->", cc.receiver)
xy = topo.nodes["QuantumRouter"][0].xy_routing

for node in topo.nodes["QuantumRouter"]:
    print(node.name, "->", xy.get_neighbors(node))
xy = topo.nodes["QuantumRouter"][0].xy_routing

print("Next hop from R0 to R8:",
      xy.next_hop("R8"))

timeline = topo.tl

r0 = timeline.get_entity_by_name("R0")

routing = r0.network_manager.get_routing_protocol()

print("R0 routing protocol:", type(routing).__name__)

routing.test_route("R8")

```
