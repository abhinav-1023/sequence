```python
from enum import Enum

class ClassicalMsgType(Enum):
    DATA = 0


class ClassicalMessage(Message):

    def __init__(self, sender, receiver, payload):

        super().__init__(ClassicalMsgType.DATA, receiver)

        self.sender = sender
        self.receiver = receiver
        self.payload = payload

    def __str__(self):

        return (
            f"ClassicalMessage("
            f"sender={self.sender}, "
            f"receiver={self.receiver}, "
            f"payload={self.payload})"
        )
```
